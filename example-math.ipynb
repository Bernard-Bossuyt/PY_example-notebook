```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

## 1. NumPy — Generate a random matrix and basic operations

```python
np.random.seed(42)

# Generate a 5x5 random matrix
matrix = np.random.randint(1, 100, size=(5, 5))
print("Random 5x5 Matrix:")
print(matrix)

print(f"\nDeterminant:  {np.linalg.det(matrix):.2f}")
print(f"Trace:        {np.trace(matrix)}")
print(f"Mean:         {matrix.mean():.2f}")
print(f"Std Dev:      {matrix.std():.2f}")

# Eigenvalues
eigenvalues, _ = np.linalg.eig(matrix)
print(f"\nEigenvalues:  {np.round(eigenvalues, 2)}")
```

## 2. Pandas — Fake sales data as a DataFrame

```python
# Generate fake sales data
n = 200
df = pd.DataFrame({
    "region":   np.random.choice(["Belgium", "France", "Germeny"], n),
    "product":  np.random.choice(["Panel A", "Panel B", "Panel C"], n),
    "units":    np.random.randint(10, 500, n),
    "price":    np.round(np.random.uniform(50, 300, n), 2),
})
df["revenue"] = df["units"] * df["price"]

print(df.head(10))
print("\n--- Summary by Region ---")
df.groupby("region")[["units", "revenue"]].agg(["sum", "mean"]).round(2)
```

## 3. Matplotlib — 3D Surface Plot

```python
# Create a 3D surface from a mathematical function
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2)) * np.exp(-0.1 * (X**2 + Y**2))

fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection="3d")
surf = ax.plot_surface(X, Y, Z, cmap="viridis", edgecolor="none", alpha=0.9)
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
ax.set_title("Damped Ripple:  z = sin(√(x² + y²)) · e^(-0.1(x² + y²))")
fig.colorbar(surf, shrink=0.5, aspect=10)
plt.tight_layout()
plt.show()
```
